### Import libraries

```python
import numpy as np                                          # numerical computing and array manipulation

from tensorflow.keras.preprocessing.text import one_hot     # converts words into integer indices
from tensorflow.keras.layers import Embedding               # converts word indices into dense vectors
from tensorflow.keras.models import Sequential              # builds a model with sequential layers
from tensorflow.keras.utils import pad_sequences            # makes all sequences the same length
```

```python
samples = ['the glass of milk','the glass of juice','the cup of tea','I am a good by',
           'I am a good developer','understand the meaning of word','your videos are good']
```

```python
# Define vocabulary size
voc_size = 10000
```

```python
samples
```

### One-Hot Representation

**NOTE**: `one_hot(input_test, n, filters, lower, split, analyzer)`

```python
# One hot representation for every word
oh_repr = [one_hot(word, voc_size) for word in samples]
oh_repr
```

**Representation**  
$\Rightarrow$ [5562, 9855, 4561, 4996] are the indices of each word vector  
- `the` = 5562  
- `glass` = 9855  
- `of` = 4561  
- `mile` = 4996 

> After dimensionality reduction using PCA, some vectors are very near to each other after plotting using **cosine similarity**.

**NOTE:**  
- Since the length of the senentes are different, maximum length are needed to be set the length before training.
- We set the maximum length of the words as `8`, but the real maximum length is `5`. We need to set the extra 3 values as 0 before or after the real words using `pad_sequences()`.

```python
# Word embedding representation
max_length = 8
embedded_docs = pad_sequences(pad_sequences=oh_repr, padding='pre', maxlen=max_length)   # Pre insertion (USED)
print(embedded_docs)
print(pad_sequences(pad_sequences=oh_repr, padding='post', maxlen=max_length))           # Post insertion
```

```python
# Feature representation (Embedding model)
dim = 10

model = Sequential()
model.add(Embedding(input_dim=voc_size, output_dim=dim, input_shape=(max_length, )))
model.compile(optimizer='adam', loss='mse', metrics=['mse'])
model.summary()
```

```python
model.predict(embedded_docs)
# NOTE: Every words represent 10 dimensions.
```
